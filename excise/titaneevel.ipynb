```python
import torch
import numpy as np
import pandas as pd
```

```python
test = pd.read_csv('./dataset/titanic/test.csv')
```

```python
test.isnull().sum()
test = test.drop(['Cabin','Embarked','Name','Ticket'],axis = 1)
```

```python
test['Age'] = test['Age'].fillna(test['Age'].mean())
test['Fare'] = test['Fare'].fillna(test['Fare'].mean())
test.isnull().sum()
gender_map = {
            'male':0,
            'female':1
        }
test['Sex'] = test['Sex'].map(gender_map)
```

```python
test['Survived'] = np.nan
test.head()
```

```python
net = torch.load('./titanic.pth')
for i in range(len(test)):
    x = torch.tensor(test.iloc[i, 1:7].values,dtype = torch.float32)
    y = net(x)
    if y < 0:
        test.iloc[i,7] = 0
    else:
        test.iloc[i,7] = 1
```

```python
out = test[['PassengerId','Survived']]
out.to_csv('./output.csv', index = False)
```
